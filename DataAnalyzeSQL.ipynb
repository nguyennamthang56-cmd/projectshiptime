```python
# CELL: CREATE SQLITE DATABASE FROM CSV FILE

import pandas as pd
import sqlite3

# Load dataset
df = pd.read_csv("dynamic_supply_chain_logistics_dataset.csv")

# Create SQLite database
conn = sqlite3.connect("logistics.db")

# Save DataFrame to SQLite table
df.to_sql("logistics", conn, if_exists="replace", index=False)

print("Dataset loaded successfully!")
print("SQLite database created successfully!")
print("Dataset shape:", df.shape)
```

```python
# CELL: SQL QUERY FUNCTION

def run_query(query):
    return pd.read_sql_query(query, conn)
```

```python
query = """
SELECT 
    timestamp,
    traffic_congestion_level,
    route_risk_level,
    weather_condition_severity,
    delay_probability,
    risk_classification,
    delivery_time_deviation
FROM logistics
ORDER BY delivery_time_deviation DESC
LIMIT 10;
"""

top_deviation = run_query(query)
display(top_deviation)
```

```python
query = """
SELECT 
    risk_classification,
    COUNT(*) AS total_records,
    AVG(delivery_time_deviation) AS avg_delivery_deviation,
    MAX(delivery_time_deviation) AS max_delivery_deviation
FROM logistics
GROUP BY risk_classification
ORDER BY avg_delivery_deviation DESC;
"""

risk_analysis = run_query(query)
display(risk_analysis)
```

```python
query = """
SELECT 
    ROUND(traffic_congestion_level, 0) AS traffic_level_group,
    COUNT(*) AS total_records,
    AVG(delivery_time_deviation) AS avg_delivery_deviation
FROM logistics
GROUP BY ROUND(traffic_congestion_level, 0)
ORDER BY traffic_level_group;
"""

traffic_analysis = run_query(query)
display(traffic_analysis)
```

```python
query = """
SELECT 
    ROUND(route_risk_level, 0) AS route_risk_group,
    COUNT(*) AS total_records,
    AVG(delivery_time_deviation) AS avg_delivery_deviation
FROM logistics
GROUP BY ROUND(route_risk_level, 0)
ORDER BY route_risk_group;
"""

route_risk_analysis = run_query(query)
display(route_risk_analysis)
```

```python
query = """
SELECT 
    timestamp,
    traffic_congestion_level,
    route_risk_level,
    delay_probability,
    risk_classification,
    delivery_time_deviation
FROM logistics
WHERE delay_probability >= 0.8
ORDER BY delivery_time_deviation DESC
LIMIT 10;
"""

high_delay_records = run_query(query)
display(high_delay_records)
```

```python
query = """
SELECT 
    strftime('%Y-%m', timestamp) AS month,
    COUNT(*) AS total_records,
    AVG(delivery_time_deviation) AS avg_delivery_deviation
FROM logistics
GROUP BY month
ORDER BY month;
"""

monthly_analysis = run_query(query)
display(monthly_analysis)
```
